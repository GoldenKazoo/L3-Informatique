```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
```

**1 – Récupérer le dataset load_breast_cancer**

```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
X = data.data
y = data.target

print("Noms des classes :", data.target_names)
print("Noms des features :", data.feature_names)
```

**Taille de la matrice de données**

```python
print("Shape de X (data) :", X.shape)
# (569, 30) : 569 exemples, 30 attributs
```

**Partitionnement du dataset : 77% train / 23% test**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.23, random_state=42
)
```

**Taille des ensembles d'apprentissage et de test**

```python
print("Taille ensemble d'apprentissage :", X_train.shape[0])
print("Taille ensemble de test         :", X_test.shape[0])
```

---
## Comparaison des méthodes de classification

Dans ce TP, on compare :
- **SVM** (Support Vector Machine)
- **KNN** (k plus proches voisins)
- **SGDClassifier**
- **Arbre de Décision**

Pour chaque méthode : sans prétraitement, avec StandardScaler, avec MinMaxScaler.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import accuracy_score

# Préparation des données prétraitées (fait une seule fois, réutilisé partout)
# StandardScaler
ss = StandardScaler()
X_train_std = ss.fit_transform(X_train)
X_test_std  = ss.transform(X_test)

# MinMaxScaler
mms = MinMaxScaler()
X_train_mm = mms.fit_transform(X_train)
X_test_mm  = mms.transform(X_test)

# Dictionnaire pour stocker tous les scores
results = {}
```

---
### SVM (Support Vector Machine)

1. Sans prétraitement  
2. Avec StandardScaler  
3. Avec MinMaxScaler

```python
from sklearn.svm import SVC

# 1. Sans prétraitement
svm_raw = SVC()
svm_raw.fit(X_train, y_train)
score_svm_raw = accuracy_score(y_test, svm_raw.predict(X_test))
results['SVM (brut)'] = score_svm_raw
print(f"SVM sans prétraitement      : {score_svm_raw:.4f}")

# 2. Avec StandardScaler
svm_std = SVC()
svm_std.fit(X_train_std, y_train)
score_svm_std = accuracy_score(y_test, svm_std.predict(X_test_std))
results['SVM (StandardScaler)'] = score_svm_std
print(f"SVM + StandardScaler        : {score_svm_std:.4f}")

# 3. Avec MinMaxScaler
svm_mm = SVC()
svm_mm.fit(X_train_mm, y_train)
score_svm_mm = accuracy_score(y_test, svm_mm.predict(X_test_mm))
results['SVM (MinMaxScaler)'] = score_svm_mm
print(f"SVM + MinMaxScaler          : {score_svm_mm:.4f}")
```

---
### KNN (k plus proches voisins) — k = 2, 3, 4

1. Sans prétraitement  
2. Avec StandardScaler  
3. Avec MinMaxScaler

```python
from sklearn.neighbors import KNeighborsClassifier

for k in [2, 3, 4]:
    # 1. Sans prétraitement
    knn_raw = KNeighborsClassifier(n_neighbors=k)
    knn_raw.fit(X_train, y_train)
    s = accuracy_score(y_test, knn_raw.predict(X_test))
    results[f'KNN k={k} (brut)'] = s
    print(f"KNN k={k} sans prétraitement : {s:.4f}")

    # 2. Avec StandardScaler
    knn_std = KNeighborsClassifier(n_neighbors=k)
    knn_std.fit(X_train_std, y_train)
    s = accuracy_score(y_test, knn_std.predict(X_test_std))
    results[f'KNN k={k} (StandardScaler)'] = s
    print(f"KNN k={k} + StandardScaler   : {s:.4f}")

    # 3. Avec MinMaxScaler
    knn_mm = KNeighborsClassifier(n_neighbors=k)
    knn_mm.fit(X_train_mm, y_train)
    s = accuracy_score(y_test, knn_mm.predict(X_test_mm))
    results[f'KNN k={k} (MinMaxScaler)'] = s
    print(f"KNN k={k} + MinMaxScaler     : {s:.4f}")
    print()
```

---
### SGDClassifier

1. Sans prétraitement  
2. Avec StandardScaler  
3. Avec MinMaxScaler

```python
from sklearn.linear_model import SGDClassifier

# 1. Sans prétraitement
sgd_raw = SGDClassifier(max_iter=1000000, random_state=42)
sgd_raw.fit(X_train, y_train)
s = accuracy_score(y_test, sgd_raw.predict(X_test))
results['SGD (brut)'] = s
print(f"SGD sans prétraitement      : {s:.4f}")

# 2. Avec StandardScaler
sgd_std = SGDClassifier(max_iter=1000000, random_state=42)
sgd_std.fit(X_train_std, y_train)
s = accuracy_score(y_test, sgd_std.predict(X_test_std))
results['SGD (StandardScaler)'] = s
print(f"SGD + StandardScaler        : {s:.4f}")

# 3. Avec MinMaxScaler
sgd_mm = SGDClassifier(max_iter=1000000, random_state=42)
sgd_mm.fit(X_train_mm, y_train)
s = accuracy_score(y_test, sgd_mm.predict(X_test_mm))
results['SGD (MinMaxScaler)'] = s
print(f"SGD + MinMaxScaler          : {s:.4f}")
```

---
### Arbre de Décision — max_depth = 2, 3

1. Sans prétraitement  
2. Avec StandardScaler  
3. Avec MinMaxScaler

```python
from sklearn.tree import DecisionTreeClassifier

for depth in [2, 3]:
    # 1. Sans prétraitement
    dt_raw = DecisionTreeClassifier(max_depth=depth, random_state=42)
    dt_raw.fit(X_train, y_train)
    s = accuracy_score(y_test, dt_raw.predict(X_test))
    results[f'DT depth={depth} (brut)'] = s
    print(f"DT depth={depth} sans prétraitement : {s:.4f}")

    # 2. Avec StandardScaler
    dt_std = DecisionTreeClassifier(max_depth=depth, random_state=42)
    dt_std.fit(X_train_std, y_train)
    s = accuracy_score(y_test, dt_std.predict(X_test_std))
    results[f'DT depth={depth} (StandardScaler)'] = s
    print(f"DT depth={depth} + StandardScaler   : {s:.4f}")

    # 3. Avec MinMaxScaler
    dt_mm = DecisionTreeClassifier(max_depth=depth, random_state=42)
    dt_mm.fit(X_train_mm, y_train)
    s = accuracy_score(y_test, dt_mm.predict(X_test_mm))
    results[f'DT depth={depth} (MinMaxScaler)'] = s
    print(f"DT depth={depth} + MinMaxScaler     : {s:.4f}")
    print()
```

---
### Tableau récapitulatif – Meilleur et pire score

```python
import pandas as pd

df_results = pd.DataFrame.from_dict(results, orient='index', columns=['Score (accuracy)'])
df_results = df_results.sort_values('Score (accuracy)', ascending=False)

print("=== Classement de toutes les méthodes ===")
print(df_results.to_string())

best_method  = df_results.index[0]
worst_method = df_results.index[-1]
print(f"\n✅ Meilleure méthode : {best_method}  →  {df_results.loc[best_method, 'Score (accuracy)']:.4f}")
print(f"❌ Pire méthode      : {worst_method}  →  {df_results.loc[worst_method, 'Score (accuracy)']:.4f}")
```

---
### Matrices de confusion – SVM (exemple de référence)

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Prédictions SVM avec StandardScaler (généralement le meilleur)
y_pred_svm = svm_std.predict(X_test_std)
print("Matrice de confusion – SVM + StandardScaler :")
print(confusion_matrix(y_test, y_pred_svm))

fig, ax = plt.subplots(figsize=(5, 4))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_svm,
    display_labels=data.target_names,
    ax=ax
)
ax.set_title('SVM + StandardScaler')
plt.tight_layout()
plt.show()
```

### Matrices de confusion – meilleure et pire méthode

```python
# On reconstruit les prédictions pour la meilleure et la pire méthode
# en identifiant dynamiquement quel modèle utiliser.

# Dictionnaire modèle -> (modèle entraîné, X_test correspondant)
model_map = {
    'SVM (brut)':               (svm_raw, X_test),
    'SVM (StandardScaler)':     (svm_std, X_test_std),
    'SVM (MinMaxScaler)':       (svm_mm,  X_test_mm),
    'KNN k=2 (brut)':           (KNeighborsClassifier(n_neighbors=2).fit(X_train, y_train),          X_test),
    'KNN k=2 (StandardScaler)': (KNeighborsClassifier(n_neighbors=2).fit(X_train_std, y_train),      X_test_std),
    'KNN k=2 (MinMaxScaler)':   (KNeighborsClassifier(n_neighbors=2).fit(X_train_mm, y_train),       X_test_mm),
    'KNN k=3 (brut)':           (KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train),          X_test),
    'KNN k=3 (StandardScaler)': (KNeighborsClassifier(n_neighbors=3).fit(X_train_std, y_train),      X_test_std),
    'KNN k=3 (MinMaxScaler)':   (KNeighborsClassifier(n_neighbors=3).fit(X_train_mm, y_train),       X_test_mm),
    'KNN k=4 (brut)':           (KNeighborsClassifier(n_neighbors=4).fit(X_train, y_train),          X_test),
    'KNN k=4 (StandardScaler)': (KNeighborsClassifier(n_neighbors=4).fit(X_train_std, y_train),      X_test_std),
    'KNN k=4 (MinMaxScaler)':   (KNeighborsClassifier(n_neighbors=4).fit(X_train_mm, y_train),       X_test_mm),
    'SGD (brut)':               (sgd_raw, X_test),
    'SGD (StandardScaler)':     (sgd_std, X_test_std),
    'SGD (MinMaxScaler)':       (sgd_mm,  X_test_mm),
    'DT depth=2 (brut)':               (DecisionTreeClassifier(max_depth=2, random_state=42).fit(X_train, y_train),      X_test),
    'DT depth=2 (StandardScaler)':     (DecisionTreeClassifier(max_depth=2, random_state=42).fit(X_train_std, y_train),  X_test_std),
    'DT depth=2 (MinMaxScaler)':       (DecisionTreeClassifier(max_depth=2, random_state=42).fit(X_train_mm, y_train),   X_test_mm),
    'DT depth=3 (brut)':               (DecisionTreeClassifier(max_depth=3, random_state=42).fit(X_train, y_train),      X_test),
    'DT depth=3 (StandardScaler)':     (DecisionTreeClassifier(max_depth=3, random_state=42).fit(X_train_std, y_train),  X_test_std),
    'DT depth=3 (MinMaxScaler)':       (DecisionTreeClassifier(max_depth=3, random_state=42).fit(X_train_mm, y_train),   X_test_mm),
}

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, label, title_color in zip(
    axes,
    [best_method, worst_method],
    ['green', 'red']
):
    if label in model_map:
        clf, X_te = model_map[label]
        y_pred = clf.predict(X_te)
        ConfusionMatrixDisplay.from_predictions(
            y_test, y_pred,
            display_labels=data.target_names,
            ax=ax
        )
        ax.set_title(f'{label}\n(score={results[label]:.4f})', color=title_color)

plt.tight_layout()
plt.show()
```

### Prédiction pour une nouvelle donnée

```python
X_new = [[14, 12, 75, 495, 6.1e-02, 3.2e-02, 2.10e-02,
  1.3e-02, 1.5e-01, 7e-02, 1.7e-01, 4.80e-01, 1.4e+00, 0.868e+01,
  5.71e-03, 1.2e-02, 1.8e-02, 5.7e-03, 1.8e-02, 1.881e-03, 1.4e+01,
  1.7e+01, 9.040e+01, 5.350e+02, 9.87e-02, 1.2e-01, 1.01e-01, 5.5e-02,
  2.4e-01, 6.87e-02]]

# Prédiction avec la meilleure méthode
if best_method in model_map:
    clf_best, X_te_best = model_map[best_method]
    # Appliquer le même prétraitement que celui utilisé pour cette méthode
    if 'StandardScaler' in best_method:
        X_new_scaled = ss.transform(X_new)
    elif 'MinMaxScaler' in best_method:
        X_new_scaled = mms.transform(X_new)
    else:
        X_new_scaled = X_new
    pred_best = clf_best.predict(X_new_scaled)
    print(f"Meilleure méthode ({best_method}) : classe prédite = {data.target_names[pred_best[0]]}")

# Prédiction avec la pire méthode
if worst_method in model_map:
    clf_worst, X_te_worst = model_map[worst_method]
    if 'StandardScaler' in worst_method:
        X_new_scaled_w = ss.transform(X_new)
    elif 'MinMaxScaler' in worst_method:
        X_new_scaled_w = mms.transform(X_new)
    else:
        X_new_scaled_w = X_new
    pred_worst = clf_worst.predict(X_new_scaled_w)
    print(f"Pire méthode      ({worst_method}) : classe prédite = {data.target_names[pred_worst[0]]}")
```
