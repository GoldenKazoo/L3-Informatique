**Régression Linéaire avec plusieurs variables ("features")**

 Exemples pour bibliothèque pandas


Estimer les prix des maisons avec

**3000 sqr ft area, 3 bedrooms, 40 year old**

**2500 sqr ft area, 4 bedrooms,  5 year old**

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
```

```python
df = pd.read_csv('homeprices.csv')
df
```

**Supprimer donnée manquante (dropna)**

```python
reg = linear_model.LinearRegression()

df_clean = df.dropna()

X = df_clean.drop('price', axis='columns')
print(X)
y = df_clean.price

print(y)
reg.fit(X, y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

**Estimer le prix d'une maison avec  3000 sqr ft area, 3 bedrooms, 40 year old**

```python
reg.predict([[3000, 3, 40]])
```

```python
reg.coef_[0]*3000 + reg.coef_[1]*3 + reg.coef_[2]*40 + reg.intercept_
```

**Estimer le prix avec  2500 sqr ft area, 4 bedrooms,  5 year old**

```python
reg.predict([[2500, 4, 5]])
```

**Exercice : Prédire les salaires**

Prédire les salaires pour les données suivantes en faisant l'apprentissage à partir des données du fichier `salaires.csv` :

- **2 yr experience, 9 test score, 6 interview score**
- **12 yr experience, 10 test score, 10 interview score**

```python
# Chargement du fichier salaires.csv
df_sal = pd.read_csv('salaires.csv')
print(df_sal.head())
print("\nColonnes :", df_sal.columns.tolist())
print("\nValeurs manquantes :")
print(df_sal.isnull().sum())
```

```python
# Remplacement des valeurs manquantes par la médiane de chaque colonne
df_sal = df_sal.apply(lambda col: col.fillna(col.median()))
print(df_sal)
```

```python
# Séparation features / target
# On suppose que la dernière colonne est le salaire (target)
target_col = df_sal.columns[-1]
X_sal = df_sal.drop(target_col, axis='columns')
y_sal = df_sal[target_col]

print("Features :", X_sal.columns.tolist())
print("Target   :", target_col)
```

```python
# Entraînement du modèle
reg_sal = linear_model.LinearRegression()
reg_sal.fit(X_sal, y_sal)

print("Coefficients :", reg_sal.coef_)
print("Intercept    :", reg_sal.intercept_)
```

```python
# Prédictions
pred1 = reg_sal.predict([[2, 9, 6]])
pred2 = reg_sal.predict([[12, 10, 10]])

print(f"Salaire prédit (2 ans exp, score test 9, interview 6) : {pred1[0]:,.0f}")
print(f"Salaire prédit (12 ans exp, score test 10, interview 10) : {pred2[0]:,.0f}")
```
