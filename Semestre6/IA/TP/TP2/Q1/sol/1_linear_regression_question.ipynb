<h3 style="color:purple">Exemple de problème de prévision du prix d'une maison</h3>

Le tableau ci-dessous représente les prix actuels des maisons dans le canton en fonction de la superficie en pieds carrés

**Énoncé du problème**: les données ci-dessus permettent de créer un modèle d'apprentissage automatique capable de prédire les prix des maisons en fonction de la superficie en pieds carrés


Vous pouvez représenter les valeurs du tableau ci-dessus sous forme de nuage de points (les valeurs sont indiquées par des marqueurs rouges). Après cela, on peut tracer une ligne droite qui correspond le mieux aux valeurs du graphique.

Vous pouvez dessiner plusieurs lignes comme celle-ci mais nous choisissons celle où la somme totale des erreurs est minimale

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('homeprices.csv')
# afficher dataframe
df
```

```python
%matplotlib inline
plt.xlabel('surface')
plt.ylabel('prix')
plt.scatter(df.area, df.price, color='red', marker='+')
```

```python
price = df.drop('area', axis='columns')
price
```

```python
area = df.drop('price', axis='columns')
area
```

```python
df
```

```python
donnees = df.values
print(donnees)
```

```python
# Create linear regression object
X = donnees[:, :1]
print("X")
print(X)
y = donnees[:, 1:]
y = np.ravel(y)
print("y")
print(y)

reg = linear_model.LinearRegression()
reg1 = reg.fit(X, y)
```

```python
reg2 = reg.fit(area, price)
```

**(1) Estimer le prix  d'une maison avec area = 3300 sqr ft**

```python
reg1.predict([[3300]])
```

```python
reg2.predict([[3300]])
```

```python
reg1.coef_
```

```python
reg.intercept_
```

**Y = m * X + b (m est le coefficient  et  b est  l'intercept)**

```python
3300 * 135.78767123 + 180616.43835616432
```

**(2) Prevoir le prix d'une maison avec  area = 5000 sqr ft**

```python
reg.predict([[5000]])
```

<h3>Exercice</h3>

<p>Téléchargez le fichier "canada_per_capita_income.csv" où il y a des revenus au Canada jusqu'au 2016. Créez un modèle de régression et prédisez le revenu par habitant des citoyens canadiens en 2020, ... 2025</p>

```python
# Chargement du fichier
df_canada = pd.read_csv('canada_per_capita_income.csv')
print(df_canada.head())
print(df_canada.columns.tolist())
```

```python
# Visualisation des données
plt.figure(figsize=(10, 6))
plt.scatter(df_canada.iloc[:, 0], df_canada.iloc[:, 1], color='blue', marker='o')
plt.xlabel('Année')
plt.ylabel('Revenu par habitant (USD)')
plt.title('Revenu par habitant au Canada')
plt.grid(True)
plt.show()
```

```python
# Préparation des données
X_canada = df_canada.iloc[:, 0].values.reshape(-1, 1)  # années
y_canada = df_canada.iloc[:, 1].values                  # revenus

# Création et entraînement du modèle
reg_canada = linear_model.LinearRegression()
reg_canada.fit(X_canada, y_canada)

print("Coefficient (pente) :", reg_canada.coef_[0])
print("Intercept            :", reg_canada.intercept_)
```

```python
# Prédictions pour 2020–2025
annees_future = np.array([[2020], [2021], [2022], [2023], [2024], [2025]])
predictions_canada = reg_canada.predict(annees_future)

for annee, revenu in zip(annees_future.ravel(), predictions_canada):
    print(f"Revenu prédit en {annee} : {revenu:,.0f} USD")
```

```python
# Visualisation avec la droite de régression et les prédictions
x_range = np.linspace(df_canada.iloc[:, 0].min(), 2025, 300).reshape(-1, 1)

plt.figure(figsize=(10, 6))
plt.scatter(df_canada.iloc[:, 0], df_canada.iloc[:, 1], color='blue', label='Données historiques')
plt.scatter(annees_future, predictions_canada, color='red', marker='*', s=100, label='Prédictions 2020-2025')
plt.plot(x_range, reg_canada.predict(x_range), color='green', linewidth=2, label='Droite de régression')
plt.xlabel('Année')
plt.ylabel('Revenu par habitant (USD)')
plt.title('Revenu par habitant au Canada – Régression linéaire')
plt.legend()
plt.grid(True)
plt.show()
```
