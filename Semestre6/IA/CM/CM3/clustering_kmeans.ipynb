* "Clustering" consiste à partionner le jeu de données en groupes appellés "clusters" ou grappes (catégorie)
* On essaie de trouver des centres  de clusters
* On fixe le nombre de centres $nc$ (le nombre de clusters)
* On initialise  $nc$ centers au hasard
* L'algorithme consiste à faire les deux points dans une boucle

1- Affecter chaque point  au centre du cluster le plus proche

2- Redéfinir  chaque centre de cluster comme etant la moyenne des coordonnées des points qui lui ont été affectés
* L'algorithme se termine lorsque l'affectation des points ne varie plus

On utilise la distance Euclidienne.

Soient $a=(a_1,a_2)$  et $b=(b_1,b_2)$ 2 points avec 2 dimensions.

$\sqrt{ (a_1-b_1)^2+ (a_2-b_2)^2}$

On peut utiliser la méthode Elbow pour trouver le nombre optimal de clusters

L'idée est de minimiser la distorsion:

$  \sum_{i=1}^n \sum_{j=1}^{N_i} d_j^i $

où 

*$n$: le nombre de cluster

*$N_i$: le nombre de points  dans le cluster $i$

*$d_j^i$ est la distance du j'ième point par rapport au centre du cluster $i$

https://pythonprogramminglanguage.com/kmeans-elbow-method/

```python
from sklearn import metrics
#from scipy.spatial.distance import cdist

wcss = []

for i in range(1, 11):
    model = KMeans(n_clusters=i)
    #m_kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0)
    model.fit(X)
    wcss.append(model.inertia_)
    #wcss.append(sum(np.min(cdist(X, m_kmeans.cluster_centers_, 'euclidean'),axis=1)) / X.shape[0])
    
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```
