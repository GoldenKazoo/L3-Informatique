```python
import numpy as np

#On choisit le modèle le plus proches voisins

from sklearn.neighbors import KNeighborsClassifier

```

```python
#Définir le modèle (estimateur avec les hyperparamètres)

#model (qui est un objet) = Constructeur (hyperparamètres)
#le premier modèle est le plus n_neighbors plus proches voisins

model1 = KNeighborsClassifier(n_neighbors =1)
print(model1)
```

```python
X = iris_dataset.data

y = iris_dataset.target
```

```python
print("X \n",X[:5])

print("y \n",y[:5])
```

```python
#on fait apprentissage,  fit

kn1 = model1.fit(X, y)

```

```python
#on estime l'étiquette pour une nouvelle donnée , predict


x_new = np.array([ [2,3,4,1.5]])
 

#x_new = [[3.1, 5.3, 2.5, 1.5]]

#Attention à la dimension
print("shape:",x_new.shape)

print("etiquette :{}".format(kn1.predict(x_new)))
```

```python
#On change le modèle
model3 = KNeighborsClassifier(n_neighbors =3)
print(model3)
```

```python
kn3 = model3.fit(X, y)
```

```python
x_new = [ [2,3,4,1.5]]

print(kn3.predict(x_new))
```

#### Entrainer et Tester les données

Il existe une fonction pour mélanger de données.

train_test_split

Une règle de base:

Par defaut,75 % pour l'ensemble d'apprentissage  et 25 % pour tester

Convention:
X pour données
y pour les étiquettes

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'], iris_dataset['target'], test_size=0.22)

#X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

```python
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
#On fait apprentissage sur  les données X_train
#On évalue sur les données X_test

kn1 =model1.fit(X_train, y_train)
```

Evaluation

```python


y_e1 = kn1.predict(X_test)

print(np.mean(y_e1 == y_test))

print(kn1.score(X_test, y_test))

from sklearn.metrics import accuracy_score
print(accuracy_score(y_e1,y_test))

print(y_e1)

print(y_test)
```

```python
kn3 =model3.fit(X_train, y_train)
y_e3 = kn3.predict(X_test)

print(kn3.score(X_test, y_test))
```

```python
y_e3 = model3.predict(X_test)



print(kn3.score(X_test, y_test))


print(y_e3)

print(y_test)
```

Comme les modèles ne sont pas les mêmes, les scores ne sont pas obligatoirement les mêmes
